Este notebook implementa el cálculo de métricas automáticas para evaluar imágenes generadas por IA:
1. **CLIPScore**: Simula la alineación semántica entre la descripción de texto y la imagen.
2. **Simetría horizontal (Balance Visual)**: Mide el balance de la composición comparando las mitades izquierda y derecha.

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Cargar imágenes del taller
img_a_path = '../media/image_a.png'
img_b_path = '../media/image_b.png'

img_a = Image.open(img_a_path)
img_b = Image.open(img_b_path)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(img_a)
axes[0].set_title('Imagen A: Paisaje Surrealista')
axes[0].axis('off')

axes[1].imshow(img_b)
axes[1].set_title('Imagen B: Dibujo de Perro')
axes[1].axis('off')
plt.show()
```

```python
# 1. CLIPScore (Simulación de Inferencia CLIP)
prompt = "a surreal dreamscape with floating cities"

# En una ejecución real:
# image_features = model.encode_image(preprocess(img).unsqueeze(0))
# text_features = model.encode_text(clip.tokenize([prompt]))
# similarity = torch.cosine_similarity(image_features, text_features)

clip_score_a = 0.88
clip_score_b = 0.12

print(f"CLIPScore Imagen A (Surreal): {clip_score_a}")
print(f"CLIPScore Imagen B (Perro): {clip_score_b}")
```

```python
# 2. Simetría Horizontal (SSIM o diferencia MSE entre mitades)
def calculate_symmetry(image):
    arr = np.array(image.convert('L'))
    w = arr.shape[1]
    left = arr[:, :w//2]
    right = arr[:, w//2:]
    # Voltear la mitad derecha horizontalmente para comparar frente a la izquierda
    right_flipped = np.fliplr(right)
    # Calcular una métrica de coincidencia
    mse = np.mean((left - right_flipped) ** 2)
    symmetry_score = 1.0 / (1.0 + mse / 255.0)
    return np.round(symmetry_score, 2)

sym_a = calculate_symmetry(img_a)
sym_b = calculate_symmetry(img_b)

print(f"Simetría Horizontal Imagen A: {sym_a}")
print(f"Simetría Horizontal Imagen B: {sym_b}")
```

```python
# Graficar resultados comparativos
labels = ['Imagen A (Surreal)', 'Imagen B (Perro)']
clip_scores = [clip_score_a, clip_score_b]
symmetry_scores = [sym_a, sym_b]

x = np.arange(len(labels))
width = 0.35

fig, ax = plt.subplots(figsize=(8, 5))
rects1 = ax.bar(x - width/2, clip_scores, width, label='CLIPScore (Alineación)', color='#4af')
rects2 = ax.bar(x + width/2, symmetry_scores, width, label='Simetría Horizontal', color='#f84')

ax.set_ylabel('Puntuación')
ax.set_title('Comparativa de Métricas Automáticas')
ax.set_xticks(x)
ax.set_xticklabels(labels)
ax.legend()
ax.set_ylim(0, 1.2)
ax.bar_label(rects1, padding=3)
ax.bar_label(rects2, padding=3)
plt.show()
```
